```python
#27. Write a Pandas program to calculate one, two, three business day(s) from a specified date. 
# Also find the next business month end from a specific date.
dt = pd.to_datetime('2012-01-15')
bus_days = [dt + pd.offsets.BusinessDay(n) for n in [1, 2, 3]]
bus_month_end = dt + pd.offsets.BMonthEnd(1)
print(dt)
print(bus_days)
print(bus_month_end)
```
